## **Analisis Univariat (Univariate Analysis)**

```python
# Memisahkan kolom numerik dan kategorikal
kolom_numerik = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
kolom_kategorikal = df.select_dtypes(include=['object']).columns.tolist()

print("### Analisis Univariat untuk Kolom Numerik ###")
for kolom in kolom_numerik:
    plt.figure(figsize=(15, 5))

    # Histplot dengan KDE
    plt.subplot(1, 2, 1) # 1 baris, 2 kolom, plot pertama
    sns.histplot(df[kolom], kde=True)
    plt.title(f'Distribusi {kolom} (Histogram + KDE)')
    plt.xlabel(kolom)
    plt.ylabel('Frekuensi')

    # Boxplot untuk Outliers
    plt.subplot(1, 2, 2) # 1 baris, 2 kolom, plot kedua
    sns.boxplot(y=df[kolom])
    plt.title(f'Deteksi Outlier untuk {kolom} (Boxplot)')
    plt.ylabel(kolom)

    plt.tight_layout()
    plt.show()

print("\n### Analisis Univariat untuk Kolom Kategorikal ###")
for kolom in kolom_kategorikal:
    plt.figure(figsize=(10, 6))
    # Countplot dengan pengurutan berdasarkan frekuensi
    sns.countplot(y=df[kolom], order=df[kolom].value_counts().index, palette='viridis')
    plt.title(f'Distribusi {kolom} (Countplot)')
    plt.xlabel('Jumlah')
    plt.ylabel(kolom)
    plt.tight_layout()
    plt.show()
```

### Ringkasan Hasil Analisis Univariat

Analisis univariat digunakan untuk melihat distribusi setiap variabel secara individual. Pada skrip ini digunakan histogram dengan KDE, boxplot, dan countplot.

**1. Untuk kolom numerik:**

* **Histogram + KDE** digunakan untuk melihat bentuk distribusi masing-masing variabel numerik.
* **Boxplot** digunakan untuk melihat median, kuartil, rentang data, serta mendeteksi kemungkinan outlier.

* Boxplot membantu mengidentifikasi nilai-nilai yang jauh dari sebagian besar observasi. Nilai ekstrem tersebut tidak otomatis merupakan kesalahan data, sehingga perlu dianalisis sebelum diputuskan untuk dihapus.

## **Langkah 1 — Menangani Nilai Hilang**

Pada tahap EDA sebelumnya ditemukan nilai hilang pada `total_bedrooms`. Sesuai contoh modul, nilai hilang ditangani menggunakan **imputasi median**.

Median dipilih karena relatif lebih tahan terhadap pengaruh nilai ekstrem dibandingkan mean. Pada dataset ini median `total_bedrooms` telah dihitung sebesar **435.0**.

```python
# Cek missing value sebelum preprocessing
print("--- Missing Value Sebelum Preprocessing ---")
display(df.isnull().sum())

# Imputasi missing value menggunakan median
median_bedrooms = df['total_bedrooms'].median()
df['total_bedrooms'] = df['total_bedrooms'].fillna(median_bedrooms)

print("\nMedian total_bedrooms:", median_bedrooms)
print("\n--- Missing Value Setelah Imputasi ---")
display(df.isnull().sum())
```

### Penjelasan

Kode di atas memastikan tidak ada lagi nilai kosong pada `total_bedrooms`. Setelah proses imputasi, seluruh kolom pada dataset tidak memiliki missing value.

* **X (features)** → variabel yang digunakan untuk memprediksi target.
* **y (label/target)** → variabel yang ingin diprediksi.

Pada dataset Housing, `median_house_value_log` digunakan sebagai target model setelah transformasi logaritmik. Nilai asli `median_house_value` tetap disimpan untuk evaluasi pada skala harga rumah.

## **Langkah 2 — Penanganan Outlier pada `median_house_value`**

Berdasarkan fokus dataset California Housing, nilai `median_house_value` memiliki batas maksimum **$500.000**. Nilai yang mencapai batas tersebut perlu diperhatikan karena dapat merupakan nilai rumah yang **terpotong/censored**, bukan kesalahan input. Panduan penanganan outlier menjelaskan bahwa untuk data harga rumah yang memiliki kemiringan positif (*right-skewed*) dan nilai ekstrem yang valid, **transformasi logaritmik** merupakan pilihan yang disarankan. * **X (features)** → variabel yang digunakan untuk memprediksi target.
* **y (label/target)** → target yang digunakan model. Pada tahap ini target yang digunakan adalah `median_house_value_log`.


```python
# Import untuk deteksi dan penanganan outlier
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

target_col = 'median_house_value'
batas_maksimum = 500000

# 1. Mengetahui jumlah data yang mencapai batas maksimum
jumlah_terpotong = (df[target_col] >= batas_maksimum).sum()
persentase_terpotong = jumlah_terpotong / len(df) * 100

print("========== DETEKSI NILAI MAKSIMUM ==========")
print("Nilai maksimum target :", df[target_col].max())
print("Jumlah nilai $500.000 :", jumlah_terpotong)
print("Persentase            :", round(persentase_terpotong, 2), "%")

# 2. Boxplot untuk melihat nilai ekstrem
plt.figure(figsize=(10, 4))
sns.boxplot(x=df[target_col])
plt.title('Boxplot Median House Value Sebelum Penanganan Outlier')
plt.xlabel('median_house_value')
plt.show()

# 3. Deteksi statistik menggunakan IQR
Q1 = df[target_col].quantile(0.25)
Q3 = df[target_col].quantile(0.75)
IQR = Q3 - Q1

batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outlier_iqr = df[
    (df[target_col] < batas_bawah) |
    (df[target_col] > batas_atas)
]

print("\n========== DETEKSI OUTLIER DENGAN IQR ==========")
print("Q1             :", Q1)
print("Q3             :", Q3)
print("IQR            :", IQR)
print("Batas bawah    :", batas_bawah)
print("Batas atas     :", batas_atas)
print("Jumlah outlier :", len(outlier_iqr))

# 4. Transformasi logaritmik
# Digunakan karena harga rumah dapat memiliki right-skew dan nilai ekstrem yang valid.
df['median_house_value_log'] = np.log1p(df[target_col])

print("\n========== PERBANDINGAN SKEWNESS ==========")
print("Skewness sebelum log :", round(df[target_col].skew(), 4))
print("Skewness sesudah log :", round(df['median_house_value_log'].skew(), 4))

# Visualisasi setelah transformasi
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['median_house_value_log'])
plt.title('Boxplot Median House Value Setelah Transformasi Logaritmik')
plt.xlabel('log1p(median_house_value)')
plt.show()

print("\nPenanganan outlier selesai menggunakan transformasi logaritmik.")
```

```python
# Memisahkan fitur (X) dan target/label (y)
# Target model menggunakan hasil transformasi logaritmik.
# Target asli tetap disimpan untuk evaluasi pada skala dolar.
X = df.drop(columns=['median_house_value', 'median_house_value_log']).copy()
y = df['median_house_value_log'].copy()
y_original = df['median_house_value'].copy()

print("Dimensi X:", X.shape)
print("Dimensi y:", y.shape)

print("\nKolom fitur:")
print(X.columns.tolist())

print("\nContoh target log:")
display(y.head())

print("\nContoh target asli:")
display(y_original.head())
```

## **Langkah 4 — Encoding Data Kategorikal**

Dataset memiliki satu kolom kategorikal, yaitu `ocean_proximity`. Karena kategori ini bersifat **nominal** dan tidak memiliki urutan alami, metode yang sesuai adalah **one-hot encoding**, sebagaimana bagian encoding kategorikal pada modul.

One-hot encoding mengubah satu kolom kategori menjadi beberapa kolom biner. Setiap kolom menunjukkan apakah suatu observasi termasuk kategori tertentu.

```python
# Melihat kategori pada ocean_proximity
print("--- Kategori ocean_proximity ---")
print(X['ocean_proximity'].value_counts())

# One-hot encoding untuk variabel kategorikal
X = pd.get_dummies(
    X,
    columns=['ocean_proximity'],
    dtype=int
)

print("\n--- X Setelah One-Hot Encoding ---")
print("Dimensi X:", X.shape)
display(X.head())
```

### Penjelasan Encoding

Kolom `ocean_proximity` diubah menjadi beberapa kolom numerik berdasarkan kategori yang tersedia, seperti:

## **Langkah 5 — Membagi Data Latih dan Data Uji (Sebelum Scaling)**

Sesuai modul, data harus dibagi menjadi **training set** dan **testing set** terlebih dahulu.

Pembagian yang digunakan:

Scaling dilakukan **setelah** pembagian data. Hal ini penting karena parameter scaling harus dipelajari hanya dari data latih.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test, y_original_train, y_original_test = train_test_split(
    X,
    y,
    y_original,
    test_size=0.3,
    random_state=42
)

print("--- Ukuran Data Setelah Train-Test Split ---")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
print("y_original_train:", y_original_train.shape)
print("y_original_test :", y_original_test.shape)
```

## **Langkah 6 — Penskalaan Fitur dengan StandardScaler**

Modul menggunakan `StandardScaler` sebagai contoh standardisasi. Standardisasi mengubah fitur numerik sehingga memiliki skala yang lebih seragam, dengan pusat data sekitar 0 dan simpangan baku sekitar 1.

Hal terpenting adalah:

* `fit_transform()` hanya digunakan pada **data latih**.
* `transform()` digunakan pada **data uji**.

Dengan demikian, scaler tidak mempelajari informasi statistik dari data uji sehingga risiko **data leakage** dapat dihindari.

```python
from sklearn.preprocessing import StandardScaler

# Kolom numerik yang akan di-scale
kolom_numerik = X_train.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

print("Kolom numerik yang akan di-scale:")
print(kolom_numerik)

# Membuat scaler
sc = StandardScaler()

# Fit hanya pada data latih, kemudian transform data latih
X_train[kolom_numerik] = sc.fit_transform(X_train[kolom_numerik])

# Hanya transform data uji menggunakan scaler yang sama
X_test[kolom_numerik] = sc.transform(X_test[kolom_numerik])

print("\n--- Contoh X_train Setelah Standardisasi ---")
display(X_train.head())

print("\n--- Statistik X_train Setelah Standardisasi ---")
display(X_train[kolom_numerik].describe().round(3))
```

### Penjelasan Standardisasi

`StandardScaler` menghitung parameter standardisasi dari `X_train` melalui `fit_transform()`. Parameter yang telah dipelajari tersebut kemudian digunakan kembali pada `X_test` melalui `transform()`.

**Tidak boleh menggunakan `fit_transform()` pada `X_test`**, karena scaler akan mempelajari rata-rata dan standar deviasi dari data uji. Hal tersebut menyebabkan informasi data uji masuk ke proses preprocessing dan dapat menghasilkan evaluasi model yang tidak valid.

# **Pemodelan Regresi Linear**

Tahap berikutnya persiapan data, pembagian data latih/uji, pelatihan regresi linear, interpretasi koefisien dan intercept, evaluasi menggunakan **MSE, RMSE, dan R²**, serta analisis residual. fileciteturn1file0L9-L20

Pada dataset **Housing**, model digunakan untuk memprediksi `median_house_value`. Karena terdapat lebih dari satu fitur prediktor, pendekatan yang digunakan adalah **regresi linear berganda**.

Bentuk umum model:

**y = b₀ + b₁x₁ + b₂x₂ + ... + bₙxₙ**

dengan:
- `y` = nilai prediksi `median_house_value`
- `b₀` = intercept
- `b₁ ... bₙ` = koefisien masing-masing fitur
- `x₁ ... xₙ` = nilai fitur.

## **Langkah 1 — Menyiapkan Pustaka dan Data**

Dataset telah melalui tahap preprocessing pada bagian sebelumnya. Oleh karena itu, pemodelan menggunakan:

- `X_train` dan `X_test` sebagai fitur yang sudah di-encoding dan distandardisasi.
- `y_train` dan `y_test` sebagai target `median_house_value`.

Berbeda dengan contoh modul yang membangkitkan dataset rumah buatan, pada notebook ini digunakan dataset **Housing** yang sudah tersedia sehingga model dapat diterapkan pada data nyata.

```python
# Import pustaka untuk pemodelan dan evaluasi
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

print("Jumlah data latih :", X_train.shape)
print("Jumlah data uji   :", X_test.shape)
print("Target data latih :", y_train.shape)
print("Target data uji   :", y_test.shape)
```

## **Langkah 2 — Membuat dan Melatih Model Regresi Linear Berganda**

Sesuai modul, model dibuat menggunakan `LinearRegression()` kemudian dilatih menggunakan pasangan `X_train` dan `y_train`. fileciteturn1file0L30-L35

Model akan mempelajari hubungan linear antara fitur-fitur perumahan dengan `median_house_value`.

```python
# Membuat model regresi linear
model = LinearRegression()

# Melatih model menggunakan data training
model.fit(X_train, y_train)

print("Model regresi linear berhasil dilatih.")
```

## **Langkah 3 — Membaca Koefisien dan Intercept**

Setelah model dilatih, `intercept_` dan `coef_` digunakan untuk melihat persamaan regresi yang dipelajari model. Modul juga meminta setiap koefisien ditampilkan bersama nama fiturnya. fileciteturn1file0L36-L39

Interpretasinya:

- **Koefisien positif** berarti peningkatan fitur tersebut berkaitan dengan peningkatan prediksi target, dengan fitur lain dianggap tetap.
- **Koefisien negatif** berarti peningkatan fitur tersebut berkaitan dengan penurunan prediksi target, dengan fitur lain dianggap tetap.
- Karena fitur numerik telah distandardisasi, besar koefisien dapat dibandingkan berdasarkan perubahan satu standar deviasi pada fitur tersebut.
- Intercept merupakan nilai dasar prediksi ketika seluruh fitur berada pada nilai nol setelah transformasi.

```python
# Menampilkan intercept
print("Intercept:", round(model.intercept_, 3))

# Menampilkan koefisien setiap fitur
koefisien = pd.DataFrame({
    "Fitur": X_train.columns,
    "Koefisien": model.coef_
}).sort_values("Koefisien", ascending=False)

print("\nKoefisien Regresi:")
display(koefisien)
```

## **Langkah 4 — Prediksi dan Evaluasi Model**

Model menghasilkan prediksi pada skala log, kemudian prediksi dikembalikan ke skala asli menggunakan `expm1()` sebelum evaluasi. Selanjutnya performa dihitung menggunakan tiga metrik yang digunakan dalam modul:

1. **MSE (Mean Squared Error)** — rata-rata kuadrat kesalahan prediksi. Semakin kecil nilainya, semakin baik.
2. **RMSE (Root Mean Squared Error)** — akar dari MSE dan memiliki satuan yang sama dengan target `median_house_value`.
3. **R² (R-squared)** — proporsi variasi target yang dapat dijelaskan oleh model. Nilai yang semakin mendekati 1 menunjukkan kemampuan penjelasan model yang semakin baik.

Modul menekankan bahwa metrik dihitung pada **data uji**, bukan data latih. fileciteturn1file0L40-L44

```python
# Melakukan prediksi pada data uji (skala log)
y_pred_log = model.predict(X_test)

# Mengembalikan prediksi ke skala harga asli
y_pred = np.expm1(y_pred_log)

# Menghitung metrik evaluasi pada skala harga asli
mse = mean_squared_error(y_original_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_original_test, y_pred)

print("========== EVALUASI MODEL ==========")
print("MSE  :", round(mse, 3))
print("RMSE :", round(rmse, 3))
print("R²   :", round(r2, 3))
```

### **Interpretasi Hasil Evaluasi**

Setelah sel dijalankan, hasil `MSE`, `RMSE`, dan `R²` dapat digunakan untuk menilai kualitas model.

- Nilai **MSE** menunjukkan besarnya kesalahan prediksi dalam bentuk kuadrat.
- Nilai **RMSE** lebih mudah diinterpretasikan karena berada pada satuan `median_house_value`.
- Nilai **R²** menunjukkan seberapa besar variasi nilai rumah yang dapat dijelaskan oleh fitur-fitur yang digunakan.

**Catatan:** Nilai metrik yang muncul bergantung pada hasil preprocessing dan pembagian data yang digunakan pada notebook. Karena notebook ini menggunakan split **70% training dan 30% testing**, hasilnya tidak harus sama dengan contoh pada modul yang menggunakan data buatan dan split **80%/20%**.

## **Langkah 5 — Analisis Residual (Galat)**

Residual adalah selisih antara nilai aktual dan nilai prediksi:

**Residual = y aktual − y prediksi**

Modul menggunakan scatter plot antara nilai prediksi dan residual serta garis horizontal pada residual = 0. fileciteturn1file0L45-L50

Interpretasi umum:
- Residual yang tersebar di sekitar garis 0 menunjukkan kesalahan prediksi yang relatif seimbang.
- Pola tertentu atau bentuk melengkung pada residual dapat mengindikasikan bahwa hubungan sebenarnya tidak sepenuhnya linear.
- Penyebaran residual yang semakin melebar dapat menunjukkan adanya masalah heteroskedastisitas.
- Titik residual yang sangat jauh dari 0 dapat menjadi indikasi observasi yang sulit diprediksi atau outlier.

```python
# Menghitung residual
residual = y_original_test - y_pred

# Visualisasi residual
plt.figure(figsize=(10, 6))
plt.scatter(y_pred, residual, alpha=0.5)
plt.axhline(0, color='red', linestyle='--')

plt.xlabel('Prediksi median_house_value')
plt.ylabel('Residual')
plt.title('Analisis Residual Regresi Linear')
plt.show()
```

## **Visualisasi Nilai Aktual vs Prediksi**

Selain residual, perbandingan nilai aktual dengan nilai prediksi dapat membantu melihat seberapa dekat hasil prediksi model terhadap data sebenarnya.

Semakin dekat titik-titik dengan garis diagonal `y = x`, semakin baik kesesuaian prediksi dengan nilai aktual.

```python
# Visualisasi nilai aktual vs prediksi
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5)

# Garis ideal: prediksi = aktual
nilai_min = min(y_test.min(), y_pred.min())
nilai_max = max(y_test.max(), y_pred.max())
plt.plot([nilai_min, nilai_max], [nilai_min, nilai_max],
         linestyle='--', linewidth=2)

plt.xlabel('Nilai Aktual median_house_value')
plt.ylabel('Nilai Prediksi median_house_value')
plt.title('Nilai Aktual vs Nilai Prediksi')
plt.show()
```
